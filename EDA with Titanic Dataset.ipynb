This tutorial explains various types of graphs used in EDA and when and why to use each graph. The notebook uses Kaggle's Titanic dataset for data manipulation and **Seaborn** and **Matplotlib** for visualizations.

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams['font.size'] = 10

# Load Titanic Dataset directly from URL
url = "/kaggle/input/competitions/titanic/train.csv"
df = pd.read_csv(url)

# Quick Data Cleaning for Visualization
df['Survived_Label'] = df['Survived'].map({0: 'Died', 1: 'Survived'})
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df.head()
```

---

## 1. Bar Graphs

### **When to Use**

Bar graphs highlight differences in count or magnitude between categories. Use bar graphs to compare categorical variables or display summary metrics (like averages or sums) across discrete groups.


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scenario 1: Unimodal/Simple Count - Passenger Counts by Class
sns.countplot(data=df, x='Pclass', hue='Pclass', legend=False, palette='crest', ax=axes[0])
axes[0].set_title("Scenario 1: Passenger Count by Ticket Class (Counts)")
axes[0].set_xlabel("Passenger Class")
axes[0].set_ylabel("Number of Passengers")

# Scenario 2: Grouped Bar Chart - Survival Rate by Passenger Class & Sex
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', palette='Set2', errorbar=None, ax=axes[1])
axes[1].set_title("Scenario 2: Survival Rate by Class and Sex (Proportions)")
axes[1].set_xlabel("Passenger Class")
axes[1].set_ylabel("Survival Rate")

plt.tight_layout()
plt.show()
```

---

## 2. Pie Charts

### **When to Use**

Pie charts emphasize percentages and relative proportions. Use pie charts to display proportional composition or parts-to-a-whole relationships when you have **5 or fewer distinct categories**.


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scenario 1: Overall Passenger Class Distribution
class_counts = df['Pclass'].value_counts()
axes[0].pie(class_counts, labels=[f'Class {i}' for i in class_counts.index],
            autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#99ff99','#ffcc99'])
axes[0].set_title("Scenario 1: Proportion of Passengers by Class")

# Scenario 2: Overall Survival Ratio with Donut variation
survived_counts = df['Survived_Label'].value_counts()
axes[1].pie(survived_counts, labels=survived_counts.index, autopct='%1.1f%%',
            startangle=140, colors=['#ff9999','#66b3ff'], wedgeprops=dict(width=0.4))
axes[1].set_title("Scenario 2: Overall Survival Ratio (Donut Chart)")

plt.tight_layout()
plt.show()
```

---

## 3. Line Graphs

### **When to Use**

Line charts connect points with lines highlighting trends, acceleration, deceleration, or stability across a progression. Use line graphs when tracking changes over a continuous interval (typically time) or ordinal sequences.


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scenario 1: Average Fare Paid Across Age Groups (Ordinal Sequence)
df['AgeGroup'] = pd.cut(df['Age'], bins=range(0, 81, 10))
age_fare = df.groupby('AgeGroup', observed=True)['Fare'].mean().reset_index()
age_fare['AgeGroup_Str'] = age_fare['AgeGroup'].astype(str)

axes[0].plot(age_fare['AgeGroup_Str'], age_fare['Fare'], marker='o', color='purple', linewidth=2)
axes[0].set_title("Scenario 1: Average Fare by Age Group")
axes[0].set_xlabel("Age Bracket")
axes[0].set_ylabel("Mean Fare ($)")
axes[0].tick_params(axis='x', rotation=45)

# Scenario 2: Simulated Passenger Boarding Trend over Time
time_series_data = df.sort_values('PassengerId').copy()
time_series_data['Cumulative_Passengers'] = np.arange(1, len(time_series_data) + 1)

axes[1].plot(time_series_data['PassengerId'], time_series_data['Cumulative_Passengers'], color='teal')
axes[1].set_title("Scenario 2: Cumulative Passenger Count by Entry ID")
axes[1].set_xlabel("Passenger Sequence ID")
axes[1].set_ylabel("Cumulative Total")

plt.tight_layout()
plt.show()
```

---

## 4. Box Plots

### **When to Use**

Box plots provide a compact visual summary of five key metrics (Min, $Q_1$, Median, $Q_3$, Max) while highlighting extreme values without assuming normality. Use box plots (box-and-whisker plots) to analyze summary statistics (median, quartiles, IQR) and flag statistical outliers across one or more groups.


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scenario 1: Overall Age Summary and Outlier Detection
sns.boxplot(y=df['Age'], color='gold', ax=axes[0])
axes[0].set_title("Scenario 1: Overall Age Distribution & Outliers")
axes[0].set_ylabel("Age")

# Scenario 2: Fare Distribution across Passenger Classes
sns.boxplot(data=df, x='Pclass', y='Fare', hue='Pclass', legend=False, palette='Set3', ax=axes[1])
axes[1].set_title("Scenario 2: Fare Distribution across Passenger Classes")
axes[1].set_xlabel("Passenger Class")
axes[1].set_ylabel("Fare ($)")
axes[1].set_yscale('log')  # Log scale used due to heavy fare skew

plt.tight_layout()
plt.show()
```

---

## 5. Scatter Plots

### **When to Use**

Scatter plots make linear trends, non-linear relationships, clusters, and bivariate outliers easily noticeable. Use scatter plots to evaluate relationships, correlations, or patterns between two continuous variables.

Note: Adding color mapping (`hue`) allows introducing a third categorical variable. 


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scenario 1: Relationship Between Age and Fare
sns.scatterplot(data=df, x='Age', y='Fare', alpha=0.6, color='dodgerblue', ax=axes[0])
axes[0].set_title("Scenario 1: Age vs. Fare")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Fare ($)")

# Scenario 2: Multi-Variable Scatter Plot (Age vs. Fare by Survival Status)
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived_Label', style='Survived_Label',
                palette={ 'Died': 'red', 'Survived': 'green' }, alpha=0.7, ax=axes[1])
axes[1].set_title("Scenario 2: Age vs. Fare Colored by Survival")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Fare ($)")

plt.tight_layout()
plt.show()
```

---

## 6. Histograms

### **When to Use**

Histograms reveal the underlying shape of data distributions—such as skewness, central tendency, spread, and modal peaks. Use histograms to visualize the continuous distribution of a single quantitative variable across contiguous intervals (bins).


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scenario 1: Distribution of Passenger Ages
sns.histplot(df['Age'], bins=20, kde=False, color='skyblue', ax=axes[0])
axes[0].set_title("Scenario 1: Age Distribution of Passengers (20 Bins)")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Frequency")

# Scenario 2: Fare Distribution with Skewness
sns.histplot(df['Fare'], bins=30, kde=False, color='coral', ax=axes[1])
axes[1].set_title("Scenario 2: Fare Distribution (Demonstrating Right-Skew)")
axes[1].set_xlabel("Fare ($)")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

---

## 7. Frequency Polygon

### **When to Use**

Unlike overlapping histograms, which obscure data, frequency polygons use line segments plotted over bin midpoints. This keeps multiple distributions clear when drawn on the same axes. Use a frequency polygon to compare distributions across multiple continuous datasets or subsets.


```python
# Calculate bin edges and midpoints for Age
counts, bin_edges = np.histogram(df['Age'], bins=15)
bin_centers = 0.5 * (bin_edges[1:] + bin_edges[:-1])

# Split by Survival
survived_ages = df[df['Survived'] == 1]['Age']
died_ages = df[df['Survived'] == 0]['Age']

counts_survived, _ = np.histogram(survived_ages, bins=bin_edges)
counts_died, _ = np.histogram(died_ages, bins=bin_edges)

plt.figure(figsize=(10, 5))
plt.plot(bin_centers, counts_survived, marker='o', color='green', label='Survived', linewidth=2)
plt.plot(bin_centers, counts_died, marker='s', color='red', label='Died', linewidth=2)

plt.title("Frequency Polygon: Age Distribution by Survival Status")
plt.xlabel("Age Interval Midpoints")
plt.ylabel("Frequency (Count)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

---

## 8. Frequency Curve

### **When to Use**

By smoothing out histogram noise, frequency curves reveal the underlying probability density function of a population. Use a frequency curve (often estimated using Kernel Density Estimation or KDE) to visualize a smoothed continuous probability distribution without bin edge artifacts.


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scenario 1: Smoothed Frequency Curve overlaying a Histogram
sns.histplot(df['Age'], kde=True, stat="density", bins=20, color='mediumseagreen', ax=axes[0])
axes[0].set_title("Scenario 1: Age Histogram with Density Curve")
axes[0].set_xlabel("Age")

# Scenario 2: Comparative Density Curves (KDE) Across Categories
sns.kdeplot(data=df, x='Age', hue='Pclass', common_norm=False, palette='Set1', linewidth=2, ax=axes[1])
axes[1].set_title("Scenario 2: Smooth Age Frequency Curves by Class")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Density")

plt.tight_layout()
plt.show()
```

---

## 9. Cumulative Frequency Polygon (Ogive)

### **When to Use**

An Ogive shows cumulative progress up to 100% or the total sample size, helping answer questions like *"What percentage of passengers were under age 30?"* Use an Ogive to display the running total of frequencies across continuous intervals, making it easy to identify percentiles, medians, and totals below specific thresholds.


```python
# Calculate cumulative frequency for Age
counts, bin_edges = np.histogram(df['Age'], bins=15)
cum_counts = np.cumsum(counts)

# Cumulative relative frequency (%)
cum_relative = (cum_counts / len(df)) * 100

plt.figure(figsize=(10, 5))
plt.plot(bin_edges[1:], cum_relative, marker='D', color='darkblue', linestyle='-')

# Add reference lines for percentile analysis (e.g., median / 50th percentile)
plt.axhline(50, color='r', linestyle='--', alpha=0.7, label='50th Percentile (Median)')

plt.title("Cumulative Frequency Polygon (Ogive): Age of Passengers")
plt.xlabel("Age Boundary")
plt.ylabel("Cumulative Percentage (%)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## Chart Selection Cheat Sheet

| Chart Type | Input Data | Primary Purpose | Best Scenario |
| --- | --- | --- | --- |
| **Bar Graph** | Categorical vs. Numeric | Compare values across categories | Comparing survival rates by class |
| **Pie Chart** | Single Categorical ($\le 5$ groups) | Show parts-to-a-whole percentages | Proportions of passenger classes |
| **Line Graph** | Continuous / Time-Series | Track trends over continuous steps | Fare averages across age groups |
| **Box Plot** | Continuous across Categories | Summarize statistics and flag outliers | Spotting fare outliers per class |
| **Scatter Plot** | Two Continuous variables | Examine correlation and clustering | Assessing age vs. fare paid |
| **Histogram** | Single Continuous | Evaluate distribution shape and spread | Age or fare distributions |
| **Frequency Polygon** | Continuous across Groups | Compare distributions without overlap | Comparing survivor vs. non-survivor ages |
| **Frequency Curve** | Single/Multiple Continuous | Smooth approximation of population density | Comparing distribution density by class |
| **Ogive (Cumulative)** | Single Continuous | Determine percentiles and thresholds | Finding percent of passengers under age 30 |
